# Array Traversal

# Time Complexity: O(n)
# Space Complexity: O(1)

```python
from array import *

arr1 = array('i',[1,2,3,4,5,6])
arr2 = array('d',[1.3,1.2,2.4])

def traversel(array):
    for i in array:        # ----------> looping so O(n) time complexity
        print(i)           # ----------> printing is O(1) constant time complexity

traversel(arr1)
traversel(arr2)
```
