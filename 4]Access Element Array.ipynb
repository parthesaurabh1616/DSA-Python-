# Access Element

# Time Complexity: O(1) + O(1) + O(1) = O(1)
# Space Complexity: O(1)

```python
from array import *

arr1 = array('i', [1,2,3,4,5,6])

def accessElement(array, index):
    if index >= len(array):        # -----> if takes constant time complexity O(1)
        print("There is not any element in this index") # ---> O(1) constant time complexity O(1)

    else:
        print(array[index])           # -----> O(1) constant time complexity.

accessElement(arr1, 1)
```

```python
accessElement(arr1,8)
```

```python
accessElement(arr1,6)
```

```python
accessElement(arr1,5)
```
