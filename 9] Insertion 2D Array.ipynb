# Insertion in 2D array

# Time Complexity : O(mn)

```python
import numpy as np

# Create a 2D array

twoDArray = np.array([[11,15,10,6],[10,14,11,5],[12,17,12,8],[15,18,14,9]])
print(twoDArray)
```

```python
newTwoDArray = np.insert(twoDArray, 0, [[1,2,3,4]], axis=1)  # axis = 1 column and axis = 0 row.
print(newTwoDArray)
```

```python
newTwoDArray = np.insert(twoDArray, 0, [[1,2,3,4]], axis=0)  # axis = 1 column and axis = 0 row.
print(newTwoDArray)
```

```python
newTwoDArray = np.insert(twoDArray, 1, [[1,2,3,4]], axis=0)  # axis = 1 column and axis = 0 row.
print(newTwoDArray)
```

```python
newTwoDArray = np.append(twoDArray, [[1,2,3,4]], axis=0)  # axis = 1 column and axis = 0 row.
print(newTwoDArray)
```
