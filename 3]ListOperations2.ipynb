# List Operations

# Update / Insert - List

# Space Complexity : O(n + k)
# Time Complexity : O(n)

```python
myList = [1,2,3,4,5,6,7,8]
print(myList)
```

```python
myList[2] = 33
```

```python
print(myList)
```

```python
myList.insert(4,15)   # -----> O(n)
print(myList)
```

```python
myList.append(55)     # -----> O(1)
print(myList)
```

```python
newList = [11,12,13,14]
myList.extend(newList)   # -----> O(n)
print(myList)
```
