# 1. Create an array and traverse.

# Time Complexity : O(N) because we are traversing

```python
from array import *
```

```python
my_array = array('i',[1,2,3,4,5])

# Traversing
for i in my_array:
    print(i)
```

# 3. Append any value to the array using append() method.
