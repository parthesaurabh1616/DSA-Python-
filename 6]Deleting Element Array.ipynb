# Deletion

# For last element the time complexity will be O(1) directly removing from last.

# Time Complexity : O(n)
# Space Complexity : O(1)

```python
from array import *

arr1 = array('i',[1,2,3,4,5,6])

arr1.remove(1)        # -----> Finds 1 in O(1), shifts 5 elements → O(n)
```

```python
print(arr1)
```

```python
arr1.remove(6)
```
