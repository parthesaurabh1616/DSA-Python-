# Searching Element in Array with Linear Search

# Here we will get the index of the element

# Space Complexity: O(1)
# Time Complexity: O(n) in the worst and average case.

```python
import array

my_array1 = array.array('i',[1,2,3,4,5])

def linear_search(arr, target):
    for i in range(len(arr)):    # --> O(n) time complexity as loos iterates over elements.
        if arr[i] == target:     # --> O(n) time complexity  
            return i
    return -1

print(linear_search(my_array1,5))
```

```python
print(linear_search(my_array1,8))
```

```python
print(linear_search(my_array1,-1))
```
