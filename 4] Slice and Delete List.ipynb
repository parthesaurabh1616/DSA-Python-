# Slice / Delete Operations

# Space Complexity : O(1)

```python
myList = ['a','b','c','d','e','f']

print(myList[0:2])
```

```python
print(myList[1:])
```

```python
print(myList[:])
```

```python
myList[0:2] = ['x','y']
```

```python
print(myList)
```

```python
myList.pop(1)
```

```python
print(myList)
```

```python
print(myList.pop(2))      # ------> O(1) / O(n)
print(myList)
```

```python
del myList[2]         # ------> Time Complexity O(n)
```

```python
print(myList)
```

```python
myList.remove('c')     # ------> Time Complexity O(n)    
```
